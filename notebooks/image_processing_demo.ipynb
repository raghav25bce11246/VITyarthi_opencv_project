# Image Processing Basics with OpenCV

This notebook demonstrates the same basic operations used in the project: reading an image, converting it to grayscale, splitting BGR channels, applying blur and saving results.

```python
from pathlib import Path
import sys
import cv2
import matplotlib.pyplot as plt

# Makes the notebook work when it is opened from the notebooks folder.
PROJECT_ROOT = Path.cwd().parent if Path.cwd().name == 'notebooks' else Path.cwd()
sys.path.insert(0, str(PROJECT_ROOT))

from src.basic_ops import to_grayscale
from src.channels import channel_visuals
from src.filters import average_blur, gaussian_blur
from src.io_utils import load_image, save_image

image_path = PROJECT_ROOT / 'assets' / 'sample_image.jpg'
output_dir = PROJECT_ROOT / 'outputs'
output_dir.mkdir(exist_ok=True)
```

## 1. Read and display the image

```python
image = load_image(image_path)
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(9, 5))
plt.imshow(image_rgb)
plt.title('Original image')
plt.axis('off')
plt.show()

print('Shape:', image.shape)
```

OpenCV loads colour images in **BGR** order. Matplotlib expects **RGB**, so the image is converted only for correct display.

## 2. Convert to grayscale

```python
gray = to_grayscale(image)
save_image(gray, output_dir / 'grayscale.jpg')

plt.figure(figsize=(9, 5))
plt.imshow(gray, cmap='gray')
plt.title('Grayscale image')
plt.axis('off')
plt.show()
```

## 3. Split BGR channels

```python
visuals = channel_visuals(image)

for name, channel_img in visuals.items():
    save_image(channel_img, output_dir / f'{name}_channel.jpg')

fig, axes = plt.subplots(1, 3, figsize=(14, 4))
for ax, (name, channel_img) in zip(axes, visuals.items()):
    ax.imshow(cv2.cvtColor(channel_img, cv2.COLOR_BGR2RGB))
    ax.set_title(f'{name.title()} channel')
    ax.axis('off')
plt.tight_layout()
plt.show()
```

## 4. Apply blurring

```python
avg_blur = average_blur(image, 7)
gauss_blur = gaussian_blur(image, 7)

save_image(avg_blur, output_dir / 'average_blur.jpg')
save_image(gauss_blur, output_dir / 'gaussian_blur.jpg')

fig, axes = plt.subplots(1, 3, figsize=(15, 4))
items = [('Original', image), ('Average blur', avg_blur), ('Gaussian blur', gauss_blur)]
for ax, (title, current) in zip(axes, items):
    ax.imshow(cv2.cvtColor(current, cv2.COLOR_BGR2RGB))
    ax.set_title(title)
    ax.axis('off')
plt.tight_layout()
plt.show()
```

## 5. Save the original image using `imwrite`

```python
save_path = output_dir / 'saved_original.jpg'
save_image(image, save_path)
print('Saved to:', save_path)
```

## Observation

- Grayscale removes colour information and keeps intensity values.
- The channel images show how strongly each B, G or R component contributes to different regions.
- Blurring reduces small details and sharp intensity changes.
- Gaussian blur normally gives a smoother, more natural-looking result than a simple average blur.
